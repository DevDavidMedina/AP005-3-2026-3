La tabla dinámica se enriquece agregando una variable distributiva para la generación de columnas, separando de esta manera las sumas totales en relación con la ciudad.

```python
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], columns = ["Frequency", "City"], aggfunc="sum")
```

```python
#Refine Dataframe using Multiple Conditions AND

import pandas as pd

#insta = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/Instagram.csv')
insta = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/Instagram.csv', parse_dates = ["reg date", "last login"])
insta["gender"] = insta["gender"].astype("category")
insta["Country"] = insta["Country"].astype("category")
insta["Influencer"] = insta["Influencer"].astype("bool")

insta.head()
```

```python
#Print our Male instagram users that joined since 2016 and they are from USA

mask = (insta["gender"] == "Male") & (insta["reg date"] > "2015-12-31") & (insta["Country"] == "United States")
insta[mask]
```

Filtro sofisticado combinando secuencias. Se enlazan múltiples barreras condicionales operadas matemáticamente mediante el vínculo lógico AND marcado bajo el símbolo especial (&). Los paréntesis son de uso estricto para diferenciar y enjaular la jerarquía obligatoria que debe seguir el filtro en paralelo y producir la nueva selección del DataFrame.

```python
mask1 = insta["gender"] == "Male"
mask2 = insta["reg date"] > "2015-12-31"
mask3 = insta["Country"] == "United States"
insta[mask1 & mask2 & mask3]
```

Rediseño optimizado del concepto superior. Al declarar individualmente la lógica y el resultado evaluativo sobre cada etiqueta en orden (mask1, mask2, mask3), al final el paso crucial o ejecución simultánea solo demanda relacionar o intersectar esas sencillas piezas entre corchetes, mejorando de gran forma su legibilidad general en el bloque de código final.

```python
mask = (insta["gender"] == "male") & (insta["reg date"] > "2015-12-31")
insta[mask]
```

Cierre analítico simplificando el uso a tan sólo dos reglas estrictamente ligadas bajo el conector lógico directo.
