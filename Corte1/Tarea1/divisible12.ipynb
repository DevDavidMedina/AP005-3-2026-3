### divisible12.py
Este código imprime los números entre 100 y 300 que son divisibles por 12.

**Explicación paso a paso:**
* `for i in range(100, 301):`: Inicia un ciclo `for` con la variable $i$ que recorre los números enteros desde 100 hasta 300.
* `if (i % 12) != 0:`: Evalúa si el residuo de la división entre $i$ y 12 es diferente de cero (es decir, si no es divisible por 12).
* `continue`: Si el número no es divisible por 12, omite la ejecución del resto de las instrucciones del ciclo y salta a la siguiente iteración.
* `print(i)`: Imprime el número $i$ si es exactamente divisible por 12.


```
for i in range(100, 301):
    if (i%12) != 0:
        continue
    print(i)
```
