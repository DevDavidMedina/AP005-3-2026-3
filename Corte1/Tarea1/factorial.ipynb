### factorial.py
Este código pide al usuario un número entero positivo y calcula su factorial mediante un ciclo `for`. El proceso se repite indefinidamente en un bucle `while True`.

**Explicación paso a paso:**
* `while True:`: Inicia un bucle infinito para repetir la solicitud al usuario.
* `value = int(input("Enter a positive integer value: "))`: Solicita un número al usuario y lo convierte a entero (`int`).
* `print("Value: ", value)`: Muestra el valor ingresado.
* `a = isinstance(value, int)`: Comprueba si la variable `value` es de tipo entero y guarda el resultado booleano (`True`/`False`) en `a`.
* `if a == True and value > 0:`: Verifica que el número sea entero y mayor a cero.
* `fact = 1`: Inicializa la variable acumuladora `fact` en 1.
* `for i in range (1, value + 1):`: Recorre los números desde 1 hasta `value`.
* `fact = fact*i`: Multiplica el acumulador `fact` por el valor actual de `i` en cada paso.
* `print(f'The factorial of {value} is: ', fact)`: Imprime el resultado del cálculo del factorial ($value!$).
* `else:` / `print("Please, enter a positive integer number")`: Muestra un mensaje de advertencia si el valor no cumple las condiciones.


```
while True:

    value = int(input("Enter a positive integer value: "))
    print("Value: ", value)
    a = isinstance(value, int)
    if a == True and value > 0:
        fact = 1
        for i in range (1, value + 1):
            fact = fact*i
        print(f'The factorial of {value} is: ', fact)
    else:
        print("Please, enter a positive integer number")
```
