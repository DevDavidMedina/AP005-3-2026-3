### primos2.py
Este código solicita al usuario un número para verificar cuáles son primos desde 1 hasta ese valor. Luego pregunta al usuario mediante un bucle iterativo si desea continuar evaluando otros números.

**Explicación paso a paso:**
* `a = 1` / `value = int(input('Ingrese un valor'))`: Inicializa la variable de control `a` y pide un límite al usuario.
* `while a == 1:`: Ejecuta el bucle interactivo mientras `a` sea igual a 1.
* `for i in range(1,value+1):`: Itera evaluando divisores hasta `value`.
* `if conta == 2:` / `else:`: Muestra si el número evaluado es primo o no.
* `print('Do you want to continue?. Press 1 to do that')`: Pregunta al usuario si desea continuar.
* `a = int(input())` / `if a != 1: break`: Si la respuesta es distinta de 1, detiene el ciclo `while`.


```
a = 1
value = input('Ingrese un valor')
value = int(value)

while a == 1:
    for i in range(1,value+1):
        conta = 0
        for n in range(1, i+1):
            residue = i%n
            if residue == 0:
                conta = conta + 1

    if conta == 2:
       print(f'{i} es un primo')
       print("\n")
    else:
       print(f'{i} NOOO es un primo')
       print("\n")

    print('Do you want to continue?. Press 1 to do that')
    a = input()
    a = int(a)

    if a != 1:
        break

    value = input('Ingrese un valor')
    value = int(value)
```
