### number.py
Este código genera y muestra un gráfico de línea usando la librería `matplotlib`, trazando valores aleatorios (eje Y) frente a los números del 1 al 12 (eje X).

**Explicación paso a paso:**
* `import random`: Importa el módulo para generar números aleatorios.
* `from matplotlib import pyplot as plt`: Importa el módulo de graficación `pyplot`.
* `numbers_a = range(1, 13)`: Genera una secuencia del 1 al 12 para el eje X.
* `numbers_b = [random.randint(1, 1000) for i in range(12)]`: Crea una lista de 12 enteros aleatorios entre 1 y 1000 para el eje Y.
* `plt.plot(numbers_a, numbers_b)`: Traza la gráfica de líneas con los puntos $(X, Y)$.
* `plt.show()`: Muestra la figura generada en pantalla.


```
import random
from matplotlib import pyplot as plt

# Add your code below:
numbers_a = range(1, 13)
numbers_b = [random.randint(1, 1000) for i in range(12)]
plt.plot(numbers_a, numbers_b)
plt.show()
```
