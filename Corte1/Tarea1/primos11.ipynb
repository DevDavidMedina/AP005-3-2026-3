### primos11.py
Similar al anterior, este código encuentra los números primos iterando del 1 al 30, y añade un salto de línea adicional al imprimir cada resultado.

**Explicación paso a paso:**
* `import time` / `inicio = time.time()`: Registra el tiempo inicial.
* `for i in range(1,31):`: Recorre los números enteros del 1 al 30.
* `for n in range(1, i+1):` ... `conta`: Evalúa y cuenta los divisores exactos.
* `if conta == 2:` / `print(f'{i} es un primo')` / `print("\n")`: Si es primo, imprime el mensaje y agrega un salto de línea.
* `fin = time.time()` / `print("t = ", (fin - inicio)*1000)`: Calcula y muestra el tiempo de ejecución en milisegundos.


```
import time
inicio = time.time()

for i in range(1,31):
    conta = 0
    for n in range(1, i+1):
        residue = i%n
        if residue == 0:
            conta = conta + 1
    if conta == 2:
        print(f'{i} es un primo')
        print("\n")

fin = time.time()
print("t = ", (fin - inicio)*1000)
```
