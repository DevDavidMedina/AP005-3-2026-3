### homework.py
Este código solicita dos números, realiza su suma, verifica si son iguales y muestra el tipo de dato de cada variable.

**Explicación paso a paso:**
* `a = input("Enter a number: ")` / `a = int(a)`: Solicita un valor y lo convierte a número entero (`int`).
* `b = input("Enter b number: ")` / `b = float(b)`: Solicita otro valor y lo convierte a número de punto flotante (`float`).
* `c = a + b`: Realiza la suma entre `a` y `b`.
* `if a == b:` / `print("equal")` / `else:` / `print("Different")`: Compara si los valores numéricos son iguales o distintos.
* `print("Type of a is: ", type(a))` / `print("Type of b is: ", type(b))`: Imprime los tipos de datos de `a` y `b`.
* `print("c = ", c)`: Muestra el resultado de la suma.
* `if type(a) == type(b):` / `else:`: Compara si los tipos de datos de `a` y `b` son idénticos.


```
a = input("Enter a number: ")
a = int(a)
b = input("Enter b number: ")
b = float(b)
c = a + b

if a == b:
    print("equal")
else:
    print("Different")

print("Type of a is: ", type(a))
print("Type of b is: ", type(b))
print("c = ", c)

if type(a) == type(b):
    print("a and b are of the same type")
else:
    print("a and b are of different type")
```
