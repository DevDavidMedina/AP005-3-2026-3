### loopFor.py
Este código itera sobre la cadena 'Python', saltándose la letra 't' mediante la instrucción `continue` e imprimiendo el resto de letras con una pausa de 1 segundo.

**Explicación paso a paso:**
* `import time`: Importa la librería para funciones relativas al tiempo y pausas.
* `cadena = 'Python'`: Define la cadena de texto a recorrer.
* `for letra in cadena:`: Recorre cada carácter de la cadena.
* `if letra == 't':` / `continue`: Si el carácter actual es 't', salta el resto de la iteración.
* `print(letra)`: Imprime la letra actual.
* `time.sleep(1)`: Realiza una pausa de 1 segundo entre iteraciones.


```
import time

cadena = 'Python'

for letra in cadena:
   if letra == 't':
      continue
   print(letra)
   time.sleep(1)
```
