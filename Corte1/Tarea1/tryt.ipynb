### tryt.py
Este código evalúa la expresión lógica `7 + 8 == 11` y guarda el resultado booleano en una variable, la cual imprime (el resultado es `False`).

**Explicación paso a paso:**
* `my_cool_variable = 7 + 8 == 11`: Realiza la adición $7+8=15$ y evalúa la igualdad $15 == 11$, dando `False`.
* `print(my_cool_variable)`: Imprime el resultado booleano.


```
my_cool_variable = 7 + 8 == 11
print(my_cool_variable)
```
