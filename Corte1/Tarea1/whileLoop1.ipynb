### whileLoop1.py
Este código ejecuta un bucle `while` anidado dentro de un `for`. Incrementa e imprime la variable iteradora `i` antes de interrumpirse internamente con la sentencia `break`.

**Explicación paso a paso:**
* `for i in range(1,6):`: Inicia el ciclo `for` con $i=1$.
* `while i <= 4:`: Ciclo `while` que se ejecuta mientras $i \le 4$.
* `i += 1`: Incrementa $i$ en 1.
* `print(i)`: Muestra el valor actualizado de $i$.
* `break`: Cancela el ciclo exterior `for` en su primera iteración.


```
for i in range(1,6):
    while i <= 4:
        i += 1
        print(i)
    break
```
