### primos.py
Este código encuentra e imprime los números primos del 0 al 30, midiendo y mostrando el tiempo total de ejecución del algoritmo en milisegundos utilizando la librería `time`.

**Explicación paso a paso:**
* `import time` / `inicio = time.time()`: Registra el tiempo inicial de ejecución.
* `for i in range(0,31):`: Itera por cada número $i$ desde 0 hasta 30.
* `conta = 0`: Inicializa el contador de divisores exactos.
* `for n in range(1, i+1):`: Recorre todos los números desde 1 hasta $i$.
* `residue = i%n` / `if residue == 0: conta = conta + 1`: Cuenta los divisores con residuo igual a cero.
* `if conta == 2:` / `print(f'{i} es un primo')`: Si tiene exactamente dos divisores (1 y sí mismo), imprime que es primo.
* `fin = time.time()` / `print("t = ", (fin - inicio)*1000)`: Registra el tiempo final e imprime la diferencia en milisegundos.


```
import time
inicio = time.time()

for i in range(0,31):
    conta = 0
    for n in range(1, i+1):
        residue = i%n
        if residue == 0:
            conta = conta + 1

    if conta == 2:
        print(f'{i} es un primo')

fin = time.time()
print("t = ", (fin - inicio)*1000)
```
