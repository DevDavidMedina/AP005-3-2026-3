* `#Print our Male instagram users that joined since 2016 and they are from USA`: Comentario en inglés que describe el objetivo del filtro: mostrar usuarios masculinos de Instagram registrados desde 2016 procedentes de EE. UU.

```python
mask = (insta["gender"] == "Male") & (insta["reg date"] > "2015-12-31") & (insta["Country"] == "United States")
```

* `mask = (insta["gender"] == "Male") & (insta["reg date"] > "2015-12-31") & (insta["Country"] == "United States")`: Crea una máscara booleana combinada con el operador `&` que exige: ser de género `"Male"`, haberse registrado después del 31 de diciembre de 2015 y residir en `"United States"`.

```python
insta[mask]
```

* `insta[mask]`: Aplica la máscara booleana `mask` para filtrar y desplegar los registros que cumplen las condiciones establecidas.

```python
mask1 = insta["gender"] == "Male"
```

* `mask1 = insta["gender"] == "Male"`: Asigna a `mask1` la condición booleana que verifica si la columna `"gender"` es igual a `"Male"`.

```python
mask2 = insta["reg date"] > "2015-12-31"
```

* `mask2 = insta["reg date"] > "2015-12-31"`: Asigna a `mask2` la condición que comprueba si la fecha de registro `"reg date"` es posterior al 31 de diciembre de 2015.

```python
mask3 = insta["Country"] == "United States"
```

* `mask3 = insta["Country"] == "United States"`: Asigna a `mask3` la condición que evalúa si el país `"Country"` es estrictamente `"United States"`.

```python
insta[mask1 & mask2 & mask3]
```

* `insta[mask1 & mask2 & mask3]`: Filtra el DataFrame `insta` combinando simultáneamente las tres máscaras con el operador lógico AND (`&`).
