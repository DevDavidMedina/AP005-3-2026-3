* `df["Gender"].unique()`: Devuelve un arreglo unidimensional (array) con todos los valores únicos e irrepetibles presentes en la columna `"Gender"`.

```python
df.pivot_table(values = "Amount", index = "Gender", aggfunc="sum")
```

* `df.pivot_table(values = "Amount", index = "Gender", aggfunc="sum")`: Genera una tabla dinámica que calcula la suma total (`sum`) del monto `"Amount"` agrupado por la columna `"Gender"`.

```python
df.pivot_table(values = "Amount", index = "Gender", aggfunc="mean")
```

* `df.pivot_table(values = "Amount", index = "Gender", aggfunc="mean")`: Genera una tabla dinámica que calcula el promedio (`mean`) del monto `"Amount"` agrupado por cada género `"Gender"`.

```python
df.pivot_table(values = "Amount", index = "Menu", aggfunc="sum")
```

* `df.pivot_table(values = "Amount", index = "Menu", aggfunc="sum")`: Genera una tabla dinámica sumando los importes de la columna `"Amount"` para cada platillo del menú `"Menu"`.

```python
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], aggfunc="sum")
```

* `df.pivot_table(values = "Amount", index = ["Gender", "Menu"], aggfunc="sum")`: Crea una tabla dinámica multinivel con índice compuesto por género y platillo del menú, sumando el total facturado en `"Amount"`.

```python
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], columns = "City", aggfunc="sum")
```

* `df.pivot_table(values = "Amount", index = ["Gender", "Menu"], columns = "City", aggfunc="sum")`: Construye una tabla dinámica con índices en filas (`Gender`, `Menu`) y columnas distribuidas por ciudad (`City`), totalizando la suma de `"Amount"`.

```python
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], columns = ["Frequency", "City"], aggfunc="sum")
```

* `df.pivot_table(values = "Amount", index = ["Gender", "Menu"], columns = ["Frequency", "City"], aggfunc="sum")`: Crea una tabla dinámica compleja con filas indexadas por `Gender` y `Menu` y columnas jerárquicas cruzadas por `Frequency` y `City`, sumando `"Amount"`.
