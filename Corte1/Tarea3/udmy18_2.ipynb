```python
students_rec = students.merge(grades, on = 'student_id')
```

* `students_rec = students.merge(grades, on = 'student_id')`: Fusiona los DataFrames `students` y `grades` evaluando la clave `student_id` y asigna el reporte a `students_rec`.

```python
students_rec.head()
```

* `students_rec.head()`: Muestra las 5 primeras filas del DataFrame combinado `students_rec`.

```python
students_rec1 = students_rec.merge(courses, on = 'course_id')
```

* `students_rec1 = students_rec.merge(courses, on = 'course_id')`: Fusiona el DataFrame previo `students_rec` con la tabla `courses` utilizando la clave `course_id` y guarda el resultado en `students_rec1`.

```python
students_rec1.head()
```

* `students_rec1.head()`: Muestra las primeras 5 filas del DataFrame multicombinado `students_rec1`.

```python
reg_course = students_rec1.groupby('course_name')
```

* `reg_course = students_rec1.groupby('course_name')`: Agrupa los registros del DataFrame `students_rec1` tomando como criterio la columna del nombre del curso `course_name`.

```python
reg_course.size().sort_values()
```

* `reg_course.size().sort_values()`: Calcula la cantidad de elementos en cada grupo de cursos con `.size()` y ordena el resultado de menor a mayor con `.sort_values()`.

```python
students_records = students_rec1.groupby(['course_name', 'last_name', 'first_name', 'grade'])
```

* `students_records = students_rec1.groupby(['course_name', 'last_name', 'first_name', 'grade'])`: Crea una agrupación multinivel jerárquica evaluando consecutivamente las columnas `course_name`, `last_name`, `first_name` y `grade`.

```python
students_records.size()
```

* `students_records.size()`: Calcula y presenta el conteo dimensional de registros agrupados según la jerarquía de las cuatro columnas indicadas.
