* `customers.merge(sales, on = 'cust_id').head(2)`: Aplica el método `.merge()` desde el objeto `customers` para fusionarlo con `sales` mediante la columna `cust_id` y visualiza únicamente sus 2 primeras filas con `.head(2)`.

```python
customers.merge(sales, on = 'cust_id').merge(products, on = 'prod_id').head(2)
```

* `customers.merge(sales, on = 'cust_id').merge(products, on = 'prod_id').head(2)`: Encadena dos uniones consecutivas: primero une `customers` con `sales` por `cust_id`, luego une el resultado con `products` por `prod_id`, y muestra las 2 primeras filas con `.head(2)`.

```python
customers.merge(sales, on = 'cust_id').merge(products[['prod_id', 'price']], on = 'prod_id').head(2)
```

* `customers.merge(sales, on = 'cust_id').merge(products[['prod_id', 'price']], on = 'prod_id').head(2)`: Encadena uniones uniendo primero con `sales` y luego con un subconjunto de `products` (`prod_id` y `price`), mostrando las 2 primeras filas.
