```python
import pandas as pd
```

* `import pandas as pd`: Importa la biblioteca de análisis de datos Pandas asignándole el alias estándar `pd`.

```python
data = {"name": ["Sally", "Jhon", "Kim"], "age": [4, 6 ,7], "gender": ["female", "male", "male"]}
```

* `data = {"name": ["Sally", "Jhon", "Kim"], "age": [4, 6 ,7], "gender": ["female", "male", "male"]}`: Crea un diccionario en la variable `data` donde cada clave será una columna y cada valor asociado es la lista con los datos correspondientes.

```python
pd.DataFrame(data)
```

* `pd.DataFrame(data)`: Llama al constructor `pd.DataFrame()` para convertir la estructura contenida en la variable `data` en un DataFrame de Pandas.

```python
pd.DataFrame(data, index = ["Kid1", "Kid2", "Kid3"])
```

* `pd.DataFrame(data, index = ["Kid1", "Kid2", "Kid3"])`: Crea un DataFrame a partir del diccionario `data` asignando etiquetas personalizadas al índice de filas mediante el parámetro `index`.
