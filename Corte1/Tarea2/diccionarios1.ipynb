```
num_exercises = {"functions": 10, "syntax": 13, "control flow": 15, "loops": 22, "lists": 19, "classes": 18, "dictionaries": 18}
```

Se vuelve a crear el diccionario de ejercicios por tema.

```
total_exercises = 0
```

Se crea una variable acumuladora que empieza en 0, para ir sumando los ejercicios.

```
for exercises in num_exercises.values():
    total_exercises += exercises
```

Se recorre cada valor (cantidad de ejercicios) del diccionario y se va sumando al acumulador total_exercises.

```
print(total_exercises)
```

```
biggest_brands = {"Apple": 184, "Google": 141.7, "Microsoft": 80, "Coca-Cola": 69.7, "Amazon": 64.8}
```

Se crea un diccionario con el valor de mercado (en miles de millones de dolares) de varias marcas.

```
for company, value in biggest_brands.items():
    print(company + " has a value of " + str(value) + " billion dollars. ")
```

.items() devuelve pares (clave, valor). Aqui se desempacan directamente en company y value, y se arma una frase concatenando el nombre de la marca con su valor convertido a texto con str().

```
pct_women_in_occupation = {"CEO": 28, "Engineering Manager": 9, "Pharmacist": 58, "Physician": 40, "Lawyer": 37, "Aerospace Engineer": 9}
```

Se crea un diccionario con el porcentaje de mujeres en distintas ocupaciones.

```
for occupation, percentage in pct_women_in_occupation.items():
    print("Women make up " + str(percentage) + " percent of " + occupation + "s.")
```

Igual que el bucle anterior: se recorre cada par (ocupacion, porcentaje) con .items() y se arma una frase para cada uno.
