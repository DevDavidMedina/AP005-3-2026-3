```
names = ['Jenny', 'Alexus', 'Sam', 'Grace']
```

Se crea una lista con nombres de estudiantes.

```
heights = [61, 70, 67, 64]
```

Se crea una lista con las estaturas (en pulgadas) de esos mismos estudiantes, en el mismo orden.

```
zipStudents = zip(names, heights)
```

La funcion zip() combina las dos listas, emparejando cada nombre con la estatura que esta en la misma posicion.

```
print("zipStudents: ", zipStudents)
```

Se imprime el objeto zip. Como zip() no muestra su contenido directamente, esto imprime algo como <zip object at ...>, no los pares en si.

```
students = {key: value for key, value in zip(names, heights)}
```

Esto es una dict comprehension: recorre los pares que arma zip() y construye un diccionario nuevo, usando cada nombre como clave y cada estatura como valor.

```
print(students)
```

Se imprime el diccionario ya construido: {'Jenny': 61, 'Alexus': 70, 'Sam': 67, 'Grace': 64}.

```
drinks = ["espresso", "chai", "decaf", "drip"]
```

Se crea una lista con nombres de bebidas.

```
caffeine = [64, 40, 0, 120]
```

Se crea una lista con la cantidad de cafeina de cada bebida, en el mismo orden.

```
zipped_drinks = zip(drinks, caffeine)
```

Se combinan ambas listas en pares (bebida, cafeina) usando zip().

```
print(zipped_drinks)
```

Se imprime el objeto zip, no su contenido (por eso no se ven los pares directamente).

```
drinks_to_caffeine = {key: value for key, value in zipped_drinks}
```

Se construye un diccionario a partir de los pares, usando cada bebida como clave y su cafeina como valor.

```
print(drinks_to_caffeine)
```

Se imprime el diccionario final de bebidas y su cafeina.

```
songs = ["Like a Rolling Stone", "Satisfaction", "Imagine", "What's Going On", "Respect", "Good Vibrations"]
```

```
playcounts = [78, 29, 44, 21, 89, 5]
```

```
plays = {key: value for key, value in zip(songs, playcounts)}
```

Se imprime el diccionario recien creado.

Como 'Respect' ya existia, esto no la duplica: actualiza su valor de reproducciones a 94.

```
print("After: ", plays)
```

Se imprime el diccionario ya actualizado, con la cancion nueva y el valor de 'Respect' corregido.
